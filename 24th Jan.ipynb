```python
from sklearn import tree
salary_age = [[25, 30000],[30, 40000],[35, 60000],[40, 80000],[22, 20000],[45, 90000]]
purchased = [0,0,1,1,0,1]
model = tree.DecisionTreeClassifier()
model.fit(salary_age,purchased)
print("Model is trained fullyy")
prediction=model.predict([[60,35000]])
print(prediction)
```

```python
from sklearn.linear_model import LinearRegression 
import numpy as np
import pandas as pd
x = np.array([[1],[2],[3],[4],[5]])
y = np.array([30000, 35000,40000,450000,50000])
model = LinearRegression()
model.fit(x,y)
print("Model is trained fullyy")
prediction=model.predict([[6]])
print(prediction)
```
